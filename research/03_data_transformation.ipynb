```python
from dataclasses import dataclass
from pathlib import Path
```

```python
@dataclass(frozen=True)
class DataTransformaionConfig:
    root_dir : Path
    data_path : Path
```

```python
from mlproject.constants import *
from mlproject.utils.common import read_yaml , create_directories
```

```python
class ConfigurationManager:
    def __init__(
        self,
        config_filepath = CONFIG_FILE_PATH,
        params_filepath = PARAMS_FILE_PATH,
        schema_filepath = SCHEMA_FILE_PATH
    ):
        self.config = read_yaml(config_filepath)
        self.parmas = read_yaml(params_filepath)
        self.schema = read_yaml(schema_filepath)


        create_directories([self.config.artifacts_root])



    def get_data_transformaion_config(self)->DataTransformaionConfig:
        config = self.config.data_transformation

        create_directories([config.root_dir])

        data_tranformaion_config = DataTransformaionConfig(
            root_dir=config.root_dir,
            data_path=config.data_path
        )

        return data_tranformaion_config
```

```python
import os
from mlproject import logger
from sklearn.model_selection import train_test_split
import pandas as pd
```

```python
class DataTrnasformation:
    def __init__(self , config : DataTransformaionConfig):
        self.config = config


    def train_test_splitting(self):
        data = pd.read_csv(self.config.data_path , sep=";")


        train , test = train_test_split(data)

        train.to_csv(os.path.join(self.config.root_dir , "train.csv") , index=False)
        test.to_csv(os.path.join(self.config.root_dir , "test.csv") , index=False)

        logger.info("Splitted the data into training and test set")
        logger.info(train.shape)
        logger.info(test.shape)

        print(train.shape)
        print(test.shape)
```

```python
try:
    config = ConfigurationManager()
    data_transformation_config = config.get_data_transformaion_config()
    data_transformation = DataTrnasformation(config=data_transformation_config)
    data_transformation.train_test_splitting()
except Exception as e:
    raise e
```
